```python
from Bio.PDB.PDBParser import PDBParser
import os

pdb_directory = '/content/drive/My Drive/PDB'
parser = PDBParser(QUIET=True)
pdb_data = {}
file_count = 0

for pdb_file in os.listdir(pdb_directory):
    if pdb_file.endswith('.pdb') and file_count < 10000:
        file_path = os.path.join(pdb_directory, pdb_file)
        try:
            struct = parser.get_structure(pdb_file[:-4], file_path)
            pdb_data[pdb_file[:-4]] = struct
            file_count += 1
        except ValueError as e:
            print(f"Error processing {pdb_file}: {e}")
```

```python
import mini3di
import pathlib
encoder = mini3di.Encoder()
```

```python
for pdb , struct in pdb_data.items():
  for chain in struct.get_chains():
    states = encoder.encode_chain(chain)
    sequence = encoder.build_sequence(states)
    print(pdb, sequence)
```
