```python
import time
import logging

# Configure logging
logging.basicConfig(level=logging.WARNING)
logger = logging.getLogger()

def download_af_struct(uniprot_id, fails_count=0, max_fails_count=3):
    try:
        URL = f"https://alphafold.ebi.ac.uk/files/AF-{uniprot_id}-F1-model_v3.pdb"
        response = requests.get(URL)
        if response.status_code == 200:
            with open(root_af / f"{uniprot_id}.pdb", "wb") as file:
                file.write(response.content)
        else:
            logger.warning(f"Failed to download file; HTTP status code: {response.status_code}")
            raise Exception("HTTP Error")
    except Exception as e:
        logger.warning(f"Error downloading AlphaFold2 structure for {uniprot_id}: {str(e)}")
```

**Counting number of files**

```python
from pathlib import Path

def count_files(directory):
    # Path object pointing to the directory
    path = Path(directory)

    # Use a generator expression to count files
    return sum(1 for _ in path.rglob('*') if _.is_file())

# Specify the directory path
directory_path = '/content/drive/My Drive/PDB'
total_files = count_files(directory_path)
print("Total files:", total_files)
```
